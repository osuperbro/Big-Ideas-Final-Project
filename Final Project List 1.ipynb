```python
import pandas as pd
import matplotlib.pyplot as plt

crime = pd.read_csv('e967381d-d7e9-48e3-a2a2-39262f7fa5c4 (3).csv')

crime.columns = crime.columns.str.strip()

print(crime.columns)

crime_counts = (crime.groupby('neighborhood').size().sort_values(ascending=False))

print(crime_counts)

crime_counts.head(15).plot(kind='bar',figsize=(12,6),color='steelblue')

plt.title('Top 15 Neighborhoods by Number of Firearm Seizures')
plt.xlabel('Neighborhood')
plt.ylabel('Number of Incident')

plt.tight_layout()
plt.show()
```
